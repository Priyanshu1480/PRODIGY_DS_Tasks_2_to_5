```python

df.drop(['Service_year_of_vehicle', 'Defect_of_vehicle', 'Work_of_casuality', 'Fitness_of_casuality', 'Time'], axis = 1, inplace = True)

df.head()
```

```python
categorical = [i for i in df.columns if df[i].dtype=='0']
print('The categorical variables are', categorical)
```

```python
for i in categorical:
    df[i].fillna(df[i].mode()[0],inplace=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, len(categorical)*2))  # dynamic height

plotnumber = 1

for col in categorical:
    if col != 'Pedestrian_movement':
        plt.subplot(len(categorical), 1, plotnumber)
        sns.countplot(data=df, y=col)
        plt.title(col.title())
        plt.xlabel('')
        plt.ylabel('')
        plotnumber += 1

plt.tight_layout()
plt.show()
```

```python
categorical = df.select_dtypes(include='object').columns.tolist()
print(categorical)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Remove column you don't want
categorical = [col for col in categorical if col != 'Pedestrian_movement']

plt.figure(figsize=(12, len(categorical)*2))

for i, col in enumerate(categorical, 1):
    plt.subplot(len(categorical), 1, i)
    sns.countplot(data=df, y=col, order=df[col].value_counts().index)
    plt.title(col)
    plt.xlabel('')
    plt.ylabel('')

plt.tight_layout()
plt.show()
```

```python
#importing label encoing module

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

#creating a new data frame from performing the chi2 analysis

df1=pd.DataFrame()

#adding all the categorical columns except the output to new data frame

for i in categorical:
    if i!= 'Accident_severity':
        df1[i]=le.fit_transform(df[i])
```
